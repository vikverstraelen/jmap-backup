```python
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
from email.mime.image import MIMEImage
```

```python
from email import generator
def write_eml_file(msg, filename):
    filename =  filename

    with open(filename, 'w') as file:
        emlGenerator = generator.Generator(file)
        emlGenerator.flatten(msg)
```

```python
msg = MIMEMultipart("related")
alternative_part = MIMEMultipart("alternative")

msg["Some-Header"] = "Some-Value"
fp = open("attachments/fm.jpg", 'rb')
msg_image = MIMEImage(fp.read(), _subtype="jpeg")
fp.close()

msg_image.add_header('Content-ID', '5b830593-5573-4478-aec4-88d2d6fa3932')

html = """
<html><head>
<meta http-equiv=3D"Content-Type" content=3D"text/html; charset=3Diso-8859-=
1">
<style type=3D"text/css" style=3D"display:none;"> P {margin-top:0;margin-bo=
ttom:0;} </style>
</head>
<body dir=3D"ltr">
<div style=3D"font-family: Calibri, Helvetica, sans-serif; font-size: 12pt;=
 color: rgb(0, 0, 0);" class=3D"elementToProof">
simple body</div>
<div style=3D"font-family: Calibri, Helvetica, sans-serif; font-size: 12pt;=
 color: rgb(0, 0, 0);" class=3D"elementToProof">
<br>
</div>
<div style=3D"font-family: Calibri, Helvetica, sans-serif; font-size: 12pt;=
 color: rgb(0, 0, 0);" class=3D"elementToProof">
<img style=3D"max-width: 1417px;" size=3D"7195" id=3D"image_0" data-outlook=
-trace=3D"F:1|T:1" src=3D"cid:5b830593-5573-4478-aec4-88d2d6fa3932"></div>
</body>
</html>
"""

text = """
simple body

[cid:5b830593-5573-4478-aec4-88d2d6fa3932]
"""

html_part = MIMEText(html, "html")
text_part = MIMEText(text, "plain")

alternative_part.attach(text_part)
alternative_part.attach(html_part)
msg.attach(alternative_part)
msg.attach(msg_image)

msg['Subject'] = "Simple Outlook Test"
msg['From'] = "test.from@example.com"
msg['To'] = "test.to@example.com"

write_eml_file(msg, "emails/simple-outlook-test.eml")
```
